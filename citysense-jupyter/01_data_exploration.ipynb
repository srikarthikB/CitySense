```python
import os
import time
import requests
import pandas as pd

output_folder = "data/api_chunks"
os.makedirs(output_folder, exist_ok=True)

chunk_size = 50000
offset = 0
chunk_number = 0

while True:
    print(f"Downloading chunk {chunk_number + 1}...")

    params = {
        "$select": ", ".join(columns),
        "$where": "cmplnt_fr_dt >= '2020-01-01T00:00:00'",
        "$limit": chunk_size,
        "$offset": offset,
        "$order": "cmplnt_fr_dt ASC"
    }

    response = requests.get(url, params=params, timeout=120)
    response.raise_for_status()

    data = response.json()

    if not data:
        print("Download complete!")
        break

    chunk_df = pd.DataFrame(data)

    chunk_path = os.path.join(
        output_folder,
        f"crime_chunk_{chunk_number:03d}.csv"
    )

    chunk_df.to_csv(chunk_path, index=False)

    print(f"Saved {len(chunk_df):,} rows → {chunk_path}")

    offset += chunk_size
    chunk_number += 1

    time.sleep(1)
```

```python
import os
import time
import requests
import pandas as pd

output_folder = "data/api_chunks"
chunk_size = 25000
offset = 1800000
chunk_number = 36

while True:
    print(f"Downloading chunk {chunk_number + 1}...")

    params = {
        "$select": ", ".join(columns),
        "$where": "cmplnt_fr_dt >= '2020-01-01T00:00:00'",
        "$limit": chunk_size,
        "$offset": offset,
        "$order": "cmplnt_fr_dt ASC"
    }

    for attempt in range(3):
        try:
            response = requests.get(url, params=params, timeout=180)
            response.raise_for_status()
            data = response.json()
            break
        except requests.exceptions.RequestException as e:
            print(f"Attempt {attempt + 1} failed: {e}")
            time.sleep(5)
    else:
        print("Stopping after 3 failed attempts.")
        break

    if not data:
        print("Download complete!")
        break

    chunk_df = pd.DataFrame(data)

    chunk_path = os.path.join(
        output_folder,
        f"crime_chunk_{chunk_number:03d}.csv"
    )

    chunk_df.to_csv(chunk_path, index=False)

    print(f"Saved {len(chunk_df):,} rows → {chunk_path}")

    offset += chunk_size
    chunk_number += 1

    time.sleep(1)
```

```python
df["cmplnt_to_dt"] = pd.to_datetime(
    df["cmplnt_to_dt"],
    errors="coerce"
)

invalid_dates = df[
    df["cmplnt_to_dt"].notna() &
    (df["cmplnt_to_dt"] < df["cmplnt_fr_dt"])
]

print("Invalid date records:", len(invalid_dates))

print(
    invalid_dates[
        ["cmplnt_fr_dt", "cmplnt_fr_tm", "cmplnt_to_dt", "cmplnt_to_tm",
         "ofns_desc", "boro_nm"]
    ].head(20)
)
```

```python
invalid_mask = (
    df["cmplnt_to_dt"].notna() &
    (df["cmplnt_to_dt"] < df["cmplnt_fr_dt"])
)

df.loc[invalid_mask, "cmplnt_to_dt"] = pd.NaT
df.loc[invalid_mask, "cmplnt_to_tm"] = pd.NA

print("Invalid dates remaining:", (
    df["cmplnt_to_dt"].notna() &
    (df["cmplnt_to_dt"] < df["cmplnt_fr_dt"])
).sum())

print("Dataset shape:", df.shape)
```

```python
print("Latitude range:", df["latitude"].min(), "to", df["latitude"].max())
print("Longitude range:", df["longitude"].min(), "to", df["longitude"].max())

outside_nyc = df[
    ~df["latitude"].between(40.4, 40.95) |
    ~df["longitude"].between(-74.3, -73.65)
]

print("\nCoordinates outside expected NYC range:", len(outside_nyc))

print(
    outside_nyc[
        ["boro_nm", "latitude", "longitude", "ofns_desc"]
    ].head(20)
)
```

```python
invalid_coord_mask = (
    ~df["latitude"].between(40.4, 40.95) |
    ~df["longitude"].between(-74.3, -73.65)
)

df.loc[invalid_coord_mask, ["latitude", "longitude"]] = pd.NA

print("Invalid coordinates remaining:", (
    ~df["latitude"].between(40.4, 40.95) |
    ~df["longitude"].between(-74.3, -73.65)
).sum())

print("Missing coordinates:")
print(df[["latitude", "longitude"]].isna().sum())
```

```python
valid_coord_mask = df["latitude"].notna() & df["longitude"].notna()

invalid_coords_remaining = df[
    valid_coord_mask &
    (
        ~df["latitude"].between(40.4, 40.95) |
        ~df["longitude"].between(-74.3, -73.65)
    )
]

print("Invalid non-missing coordinates remaining:", len(invalid_coords_remaining))
print("Missing coordinates:")
print(df[["latitude", "longitude"]].isna().sum())
```

```python
date_mismatch = (
    (df["year"] != df["cmplnt_fr_dt"].dt.year) |
    (df["month"] != df["cmplnt_fr_dt"].dt.month) |
    (df["day_of_week"] != df["cmplnt_fr_dt"].dt.dayofweek)
)

time_mismatch = (
    df["hour"] != df["cmplnt_fr_tm"].dt.hour
)

print("Date-derived column mismatches:", date_mismatch.sum())
print("Hour column mismatches:", time_mismatch.sum())
```

```python
same_day = df["cmplnt_to_dt"].eq(df["cmplnt_fr_dt"])

valid_times = (
    df["cmplnt_fr_tm"].notna() &
    df["cmplnt_to_tm"].notna()
)

time_order_issue = df[
    same_day &
    valid_times &
    (df["cmplnt_to_tm"] < df["cmplnt_fr_tm"])
]

print("Same-day end time earlier than start time:", len(time_order_issue))

print(
    time_order_issue[
        [
            "cmplnt_fr_dt",
            "cmplnt_fr_tm",
            "cmplnt_to_dt",
            "cmplnt_to_tm",
            "ofns_desc",
            "boro_nm"
        ]
    ].head(20)
)
```
